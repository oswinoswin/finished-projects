```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/all_cars_cleaned2.csv", index_col='vehicle reference')
df.head()
```

```python
df.describe()
```

## Rozkłady zmiennych

```python
df.hist(bins=20, figsize=(15, 15), edgecolor='black')
plt.title("Histogramy zmiennych numerycznych")
plt.tight_layout()
# plt.show()
plt.savefig("../plots/histogramy_zmiennych_numerycznych.png")
```

```python
categorical_features = ['Bodywork', 'Color', 'Fuel', 'Gearbox', 'Type']
```

```python
for col in categorical_features:
    df[col].value_counts().plot(kind='bar')
    plt.ylabel("Count")
    plt.xlabel("")
    plt.title(col)
    plt.xticks(rotation=45)
    plt.show()
```
